### A print függvény
A függvényeknek, mint amilyen a `print` függvény is, három fő tulajdonsága lehet:
- **visszatérési érték**: egy érték amelyet a függvény visszaad. (visszatérési érték)
- **mellékhatás**: valami, ami a függvény futása során bekövetkezik
- **argumentumok**: a függvény bemenete(i), melyek a függvénynevet követő zárójelpár között helyezkednek el és egymástól vesszővel vannak elválasztva

A `print()` függvény tulajdonságai:
- **mellékhatás**: A bemenetekből összefűzött szöveg a kimenetre íródik
- **argumentumok**: Tetszőleges számú paraméter adható meg neki, melyeket sztringgé alakít, majd szóközzel elválasztva összefűz
- **visszatérési érték**: Nincs, a kiírt érték csak mellékhatás

````python
print("három-szor kettő =", 3*2)
````

Ahogy láttuk a program először kiszámolta `3*2` értékét, majd ezt helyettesítette be szövegként. Ez azért van, mert a `3*2` kifejezést nem tettük idézőjelek közé, így a python tudta, hogy itt nem egy szövegről, hanem két szám közti műveletről van szó. 

A program végrehajtási sorrendje először a paraméterek értékét határozza meg, majd utána hívja meg a függvényt. Ez jelen esetben így néz ki a háttérben:
- `"három-szor kettő"` -> ez egy egyszerű szöveg, tehát nem csinál vele semmit
- `2*3` -> ez egy művelet, melynek kiértékeli az eredményét, ezt helyettesíti be a `print` függvény második paramétereként
- lefuttatja a `print("három-szor kettő", 6)` műveletet

A print függvénynek van további két key-word argumentuma (a függvények keyword argumentumait nevükön kell szólítani)
- **sep**: az argumentumok szeparálására szolgáló karakter/szöveg (alap esetben ez egy szóköz)
- **end**: ezzel a karakterrel/szöveggel fejezi be a kiírást a print (alapesetben `\n`, azaz új sor)

````python
print("alma, a fa alatt", end=" ")
print("nyári", "piros", "alma", sep="...")
````

## Aritmetikai Operátorok
Az aritmetikai operátorokat leginkább számok közti műveletek elvégzésére használjuk, ilyenek az összeadás, vagy a hatványozás.

Elképzelhetjük őket úgy, mint egy két bemenetű és egy kimenetű függvény, melynek két bemenet a két oldalán álló érték, kimenete pedig ennek eredménye. Például:
- az `5*7` kifejezést tekintve a `*` a szorzás operátor, melynek két bemenete az 5 és a 7, kimenete pedig ezeknek szorzata, azaz 35.

Az operátor két oldalán álló "paraméterek" az:
- `rhs`: right-hand-side azaz az operátor jobb oldalán álló paraméter
- `lhs`: left-hand-size azaz az operátor bal oldalán álló paraméter

Az alap operátorok a:
- `+` (összeadás): lhs és rhs összege
- `-` (kivonás): lhs és rhs különbsége
- `*` (szorzás): lhs és rhs szorzata
- `/` (osztás): lhs elosztva rhs-el
- `//` (egész-részes osztás): hány egésszer van meg rhs lhs-ben. Például 7//2 === 3
- `%` (modulo): lhs osztva rhs-el mennyi maradékot ad. Például 7%2 === 1
- `**` (hatványozás): lhs az rhs-ediken

````python
# Ha rhs és lhs is int, a kimenet is int lesz
print(2 ** 3)
# Ha valamelyikük float, a kimenet is float lesz
print(2 ** 3.)
print(2. ** 3)
print(2. ** 3.)
````

````python
# Az osztás eredménye mindig float
print(3/2)
# Még akkor is, ha a matematikai eredmény nem tört lenne
print(4/2)
````

Az egész részes osztás az osztás végeredményét kerekíti lefele.
- Ez pozitív számok esetén azt mutatja, hogy hány egésszer van meg rhs, lhs-ben.
- Negatív számok esetén egy kicsit más az eredmény, ugyanis ott a lefele kerekítés ellenkezőleg működik

````python
# Az egész részes osztás alapvetően int-tel tér vissza, hisz az eredmény mindig egy egész szám
print(6 // 3)
# De ha float szerepel a műveletben, float lesz az eredmény
print(6 // 3.)
print(6. // 3)
print(6. // 3.)
# a -6/4 eredménye -1.5 ezt viszont lefele kerekítve -2-t kapunk
print(-6 // 4)
print(6. // -4)
````

A modulo az osztási maradékot adja vissza. Például a `14 % 4` eredménye 2 lesz mivel:
- `14 // 4` eredménye `3` -> ez az egész rész;
- `3 * 4` eredménye `12` -> Ez az a része lhs-nek, amiben egésszer meg van rhs;
- `14 - 12` eredménye `2` -> Ez pedig a maradék.

````python
# Adj meg két tetszőleges számot (b nem lehet 0)
a = 14
b = 4
print(a%b)
print(a - (a // b) * b )
````

#### Operátorok feldolgozási sorrendje

````python
# Balról jobbra értékelődik ki
print(12 - 1 + 1)
# a szorzás magasabb rendű, mint az összeadás, vagy kivonás
print(10 + 1 * 10)
# az osztás és a szorzás közt nincs hierarchia (itt nem (10*10)/(10*10) lesz az eredmény)
print(10 * 10 / 10 * 10)
````

Amikor úgy gondolkodtunk, hogy `2 ** 2` -> `4`; `4 ** 3` -> `64` lesz, akkor matematikailag igazából ezt írtuk fel:

$(2^2)^3$

## Változók
most már tudunk létrehozni különféle adatokat és ezekkel műveleteket végezni, de mi van, ha ezeket el is szeretnénk tárolni, hogy később újra felhasználjuk őket? Megoldás: változók.

A változók egyféle névvel ellátott tárolók, melyekben értékeket tárolhatunk. Ahogy a változó (variable) név is sugallja, ezkenek értéke változtatható.

A változókat a kódot író hozza létre (deklarálja), annak nevének megadásával. A változóneveknek van néhány megkötése:
- A változónevek kis- és nagybetűket, számokat és _ karaktert tartalmazhatnak
- A változóneveknek betűvel, vagy _ karakterrel kell kezdődniük
- A változónév case sensitive: az `alma` és az `Alma` változónevek két külön változót jelentenek
- A python nyelvi elemi védettek, így ezek nem lehetnek változónevek. Ilyenek például a:

`False`, `None`, `True`, `and`, `as`, `assert`, `break`, `class`, `continue`, `def`, `del`, `elif`, `else`, `except`, `finally`, `for`, `from`, `global`, `if`, `import`, `in`, `is`, `lambda`, `nonlocal`, `not`, `or`, `pass`, `raise`, `return`, `try`, `while`, `with`, `yield`

A változónevek bár tartalmazhatnak ékezeteket, de ezek ez nem szép.

 Pythonban megegyezésszerűen a változóneveket kis betűkkel írjuk, a szavak közé pedig alulvonásokat teszünk:
`this_is_a_variable_name`

Úgy hozhatunk létre egy változót, hogy leírjuk a nevét és egy egyenlőség jellel értéket adunk neki.

````python
# Változó létrehozása a <név> = <érték> szintaxissal történik
my_variable = 123
print(my_variable)
````

````python
# Példa a változók felhasználására. Ide bármilyen személy adatait megadhatjuk
name = 'John Doe'
age = 53
print(name, "is", age, "years old.")
````

````python
# egy változó értékét akár felül is írhatjuk
var = 123
print(var)
var = "alma"
print(var)
````

Mérföld, kilóméter konverziós példa

A `print`-ben használt `round` függvény float-okat képes kerekíteni. Két paramétere van
- Az első paraméter a szám amit kerekíteni szeretnénk
- A második paraméter a kerekítés pontossága. Ez egy opcionális paraméter, ha nem adjuk meg, akkor int-té lesz kerekítve a tört. Amennyiben egy pozitív intet adunk meg, akkor ennyi tizedes jegyre lesz kerekítve az első paraméterben kapott tört. (A példában két tizedes jegyre kerekítünk)

````python
# Egy mérföld 1.6 kilóméter, így a két mértékegység váltószámai:
kilometers_per_mile = 1.6
miles_per_kilometer = 1/kilometers_per_mile

# Ezeket a mértékegységeket szernénk átváltani
kilometers = 12.25
miles = 7.38

# És így válthatjuk át
miles_to_kilometers = miles * kilometers_per_mile
kilometers_to_miles = kilometers * miles_per_kilometer

print(miles, "miles is", round(miles_to_kilometers, 2), "kilometers")
print(kilometers, "kilometers is", round(kilometers_to_miles, 2), "miles")
````

Készítsünk programot ami kiszámolja egy háromszög átfogóját

$c = \sqrt{a^2 + b^2}$

$\sqrt{x} = x^{0.5}$

----------------------

#### Elágazások: if-else

Az `if` kifejezés egy speciális python nyelvi elem. Ezzel egy feltételes futási ágat hozhatunk létre. Az `if` felépítése az alábbi:

```python
if feltétel:
    print("valami")
```

Egy feltételes utasítás a következő módon épül fel, ebben a sorrendben:
- az `if` kulcsszó;
- egy vagy több szóköz;
- egy kifejezés (ciklusfeltétel), amelynek értékét **bool**-ként lehet értelmezni;
- kettőspont (`:`) és egy új sor;
- egy behúzott (tabulátorral) utasítás vagy utasítások sora (legalább egy utasítás kötelező); a behúzás kétféleképpen érhető el:
  - egy adott számú szóköz (az ajánlás szerint **négy szóköz**);
  - vagy a tabulátor (`Tab`) karakter használatával.

Működése:
- ha a ciklusfeltétel egy `**True**` értéket reprezentál (vagy `True` az értéke, vagy bool-á konvertálva `True` lenne), a behúzott sorok kiértékelődnek

## Ciklusok
A ciklusokkal egy utasítást, vagy utasítássorozatot hajthatunk végre többször egymás után. Ennek két féle módja van:
- **`while` ciklus** – addig ismétlődik, amíg egy adott feltétel igaz marad.
- **`for` ciklus** – ismert számú ismétléshez használjuk, kifejtésére később kerül sor.

A **`while` ciklus** felépítésében hasonlít egy `if` kifejezésre. A while kulcsszó után egy bool **feltétel** következik, majd egy : karakter, ez alatt pedig behúzva a ciklustörzset alkotó sorok.
A ciklus megvizsgálja a feltételt és amennyiben az `True`-nak bizonyul, lefuttatja a ciklustörzset. Ekkor megint megvizsgálja a feltételt, és ha az még mindig igaz, akkor megint lefuttatja a ciklustörzset. Könnyen láthatjuk, hogy a ckilustörzsben valami olyannak is kell történnie, ami befolyással van a feltételre, máskülönben a ciklus örökké futna.

Készítsünk egy programot, amely:
- számokat kér be a felhasználótól
- ha a megadott szám -1, akkor nem kérünk be több számot
- a program futása végén írjuk ki a legnagyobb látott számot

````python
# Kérjük be az első számot
szam = int(input("Adj meg egy számot: "))
# A legnagyobb szám jelenleg nyilván csak ez a szám lehet
legnagyobb_szam = szam

# Addig fusson a ciklus amíg nem kaptunk -1-et
while szam != -1:
    
    if szam > legnagyobb_szam:
        legnagyobb_szam = szam
    # A ciklustörzs végén pedig bekérjük a következő számot
    szam = int(input("Adj meg egy számot: "))

print(f"A legnagyobb megadott szám: {legnagyobb_szam}")
````

Meghatározott mennyiségű lépéshez használhatunk egy számláló változót

Példa: írjuk ki 2-nek az első 10 hatványát

````python
counter = 0
while counter < 10:
    print(f"2-nek a {counter} hatványa: {2 ** counter}")
    counter += 1
````

Az ilyes fajta ciklusok nagyon gyakoriak, ezért egy külön beépített megoldás is van rájuk, ez pedig a **`for`** ciklus.

````python
for counter in range(10):
    print(f"2-nek a {counter} hatványa: {2 ** counter}")
````

A `for` ciklus felépítése:
- A `for` kulcsszóval kell kezdeni
- Ezt követi a ciklusváltozó (melynek értéke cikluslépésenként változni fog)
- Ezt követi az `in` kulcsszó, ez rendeli össze a bal oldalán álló ciklusváltozót, a jobb oldalán álló értékkészlettel, a ciklusváltozó ezeket az értékeket fogja vissza adni.
- A következő a `range()` függvény. Ez hozza létre azokat az értékeket (számok 0-tól 10-ig), melyeket a ciklusváltozó felvesz az egyes lépések során. A `range(10)` eredménye 0-tól kezdve 10-ig az összes egész számot tartalmazza, a 10-et már nem bele értve.
- Szokásosan ezt követi egy : karakter
- Majd a ciklustörzs (A for ciklus ciklustörzsében nem kell növelni a ciklusváltozót, ugyanis ez folyamatosan új értékeket vesz ki magának a `range(10)` eredményéből)

A `range` függvénynek több paramétert is adhatunk:
- `range(y)`: számok 0-tól y-ig
- `range(x, y)`: számok x-től y-ig
- `range(x, y, z)`: számok x-től y-ig z-esével
- ebből következik, hogy `range(10)` == `range(0, 10, 1)`, hiszen a számokat vesszük 0-tól 10-ig, 1-esével

````python
for i in range(2,10):
    print(i)
````

````python
for i in range(0, 10, 2):
    print(i)
````

````python
for i in range(10, 0, -2):
    print(i)
````

A **`break`** és **`continue`** speciális kifejezések, melyeket egy ciklusban használunk
- **`break`**: segítségével kitörhetünk a ciklusból
- **`continue`**: segítségével a következő cikluslépésre ugorhatunk

````python
for i in range(8):
    if i == 3:
        break
    print(i)
print("ciklus vége")
````

Amint látni, mikor `i` értéke `3` lett, kiértékelődött a `break`, amely így megszakította a ciklus futását.

````python
for i in range(8):
    if i == 3:
        continue
    print(i)
print("ciklus vége")
````

Amint látni, mikor `i` értéke `3` lett, kiértékelődött a `continue`, ezáltal a ciklus rögtön a következő lépésre lépett, így a `3`-as szám kiíratása már nem történt meg.

A for ciklusnak lehet egy `else` ága, amely csak akkor fut le, ha nem break-kel léptünk ki a ciklusból.

````python
for i in range(10):
    break
else:
    print("A ciklus break nélkül végig futott")
print("ciklus vége")
````

Mi van ha két ciklusunk van egymásba építve, és a belsőben break-elünk?

````python
for x in range(3):
    for y in range(10):
        if y == 5:
            break
        print(f"x = {x}; y = {y}")
````

Amint azt látni, a break csak a belső ciklust szakítja meg. Az eddigiek ismeretében hogyan szakítanánk meg belülről a külső ciklust?

Példafeladat:
- Egy ciklusban kérjünk be neveket
- Amikor megkapunk egy nevet, kérjünk be 4 számot
- adjuk össze a 4 számot
- Írjuk bele egy sztringbe a nevet, egy kettőspontot és az összeget
- vesszővel elválasztva tegyük bele a következő nevet és összeget
- csináljuk mindezt addíg, míg a megadott szám nem -1

````python
szoveg = ""
# A while True ciklus feltétele mindig igaz, így ez mindaddig fut, míg egy break az meg nem szakítja
while True:
    nev = input("Add meg a következő nevet: ")
    szumma = 0
    # Amikor a for ciklus ciklusváltozójának értékét nem használjuk, jellemzően "_" néven nevezzük el
    for _ in range(4):
        szam = int(input("Adj meg egy szamot: "))
        szumma += szam
        if szam == -1:
            break
    else:
        szoveg += f"{nev}: {szumma}, "
        continue
    break
print(szoveg)
````

Hogy működik ez?
- A `while True` ciklus mindaddig fut, míg meg nem szakítjuk egy `break`-kel
- A `while` ciklus minden lépésében bekérünkk egy új nevet, és a `szumma` értékét 0-ra állítjuk
- Ez után belépünk a `for` ciklusba, ami minden lépésnél bekér egy számot
- A számot ezután hozzá adja a `szumma` változó értékéhez, ami így minden név esetén 0-ról indul, majd egyesével hozzá lesznek adva a számok.
- Ha a számok közt nem szerepelt a `-1`:
    - belépünk az `else` ágba
    - itt a `szoveg` sztringhez hozzáfűzzük a nevet és az ehhez tartozó `szumma` értéket
    - majd a `continue` utasítás következik, ami által rögtön a `while` ciklus következő iterációjába kerülünk -> bekérjük az új nevet
- Ha a `szam` `-1` volt:
    - break-elünk a belső `for` ciklusból.
    - nem kerülünk be az `else` ágba, hanem megyünk tovább a `while` ciklustörzsben
    - elérünk a külső `break`-hez, ezáltal kilépünk a `while` ciklusból.

Tehát az `else` ágba rejtett `continue` utasítás miatt tudjuk elkerülni a `break` utasítást, ezáltal elértük, hogy egy belső ciklus tudja megszakítani a külsőt

A `for` ciklus nem csak a `range` által visszaadott számokon tud végig-iterálni, de akár egy szöveg karakterein is:

Alább látható egy szkript, amely mOcKInG formára alakít át egy szöveget.
- ehhez véletlenszerűen kell egy-egy betűt nagybetűsíteni
- a véletlenszerű működéshez a `random` modult fogjuk felhasználni
- a nagybetűsítés a `"valamilyen sztring".upper()` módon történik

````python
import random
szoveg = "Ma nem megyek veletek, mert estig dolgozok."
mockolt_szoveg = ""
for betu in szoveg:
    # a random.randint(0,1) vagy 0-t, vagy 1-et ad vissza
    if random.randint(0,1):
        mockolt_szoveg += betu.upper()
    else:
        mockolt_szoveg += betu
print(mockolt_szoveg)
````

Az ilyen struktúrák a listák.

````python
# Öt szám rendezése listák nélkül -> bonyolult és túl hosszadalmas.
a, b, c, d, e = 5, 232, 5, 0, 92

# Ha találunk az első elemnél kisebbet, megcseréljük őket
if a > b:
    a, b = b, a
if a > c:
    a, c = c, a
if a > d:
    a, d = d, a
if a > e:
    a, e = e, a

# Ha találunk a 3. 4. vagy 5. elemek közt a másodiknál kisebb elemet, kicseréljük vele
if b > c:
    b, c = c, b
if b > d:
    b, d = d, b
if b > e:
    b, e = e, b

# Harmadik elem
if c > d:
    c, d = d, c
if c > e:
    c, e = e, c

# Negyedik és ötödik elem
if d > e:
    d, e = e, d

# Ahogy látjuk, 5 elem rendezésénél ez 4+3+2+1 = 10 összehasonlítás, általánosan ez n*(n-1)/2 if-ág
print(a, b, c, d, e)
````

A **függvények** és metódusok nagyon hasonló nyelvi elemek, egy különbséggel. A függvények, mint a `range`, `len`, az `input`, vagy `print`, önmagukban hívhatóak, és argumentumokat várnak. Futása lényegében olyan, mint egy függvényé, lehet visszatérési értéke és mellékhatása is. Amikor egy objektumon hívunk meg egy metódust, akkor a függvény magát az objektumot is megkapja rejtetten, mint argumentum. 

A metódusok adattípusokhoz kötöttek, például:
- A sztringeknek van egy `"alma a fa alatt".split()` - `split` metódusa, amely a szóközök mentén egy listányi sztringgé darabolja fel azt a sztringet amin meg volt hívva
- A listáknak is vannak különböző metódusai, mint a `.append()`, vagy `.pop()`, ezek lesznek bemutatva

A metódusok felépítése:
- Mindig egy konkrét adat után kezdődik, egy `.` karakterrel
- ezt követi a metódus neve
- majd zárójel párok közt az argumentumok
- Vannak metódusok, amiknek visszatérési értéke van, ilyen a fent említett `split` is, és olyanok is, amelyek az adatot módosítják amelyen meg voltak hívva, ilyenek az `append` és a `pop`

Elemek megcserélése

Mit tegyünk, ha két változó értékét szeretnénk megcserélni?
Ez a megoldás biztosan nem jó:
```python
a = 1
b = 2
a = b
b = a
```
Miért nem?

Amikor azt mondtuk, hogy `a = b`, akkor `a`-nak az értéke `2`-lett, míg `b`-nek eddig is `2` volt. Ezt követően pedig a `b = a` már `2`-n hagyja mindkét értéket

Megoldás: átmeneti változó
```python
a = 1
b = 2
temp = a
a = b
b = temp
```
Ez azért működik, mert arra az időre, míg `a` felveszi `b` értékét, `a` korábbi értékét megőrizzük a `tmp` változóban. Ez egy kicsit feleslegesnek tűnik, szerencsére a pythonban van erre is nyelvi megoldás:
```python
a = 1
b = 2
a, b = b, a
```

````python
# Egyszerre több változónak is adhatunk értéket
a, b = 1, 2
print(f"a = {a}; b = {b}")
# A csere során, igazából mindkét változónak új értéket adunk, ezek éppen egymás értékei lesznek.
a, b = b, a
print(f"a = {a}; b = {b}")
````

````python
# Ez a szintaktika listáknál is működik:
l = [0,1,2,3]
print(l)

l[0], l[2] = l[2], l[0]
print(l)
````

````python
# Lista megfordítása
l = list(range(20))
print(l)

for i in range(len(l) // 2):
    l[i], l[-i - 1] = l[-i - 1], l[i]

print(l)
````

Hogy működött ez?
- a kifejezés ami érdekes az `l[i], l[-i - 1] = l[-i - 1], l[i]`
- Első esetben `i = 0`, ekkor ez a kifejezés így néz ki: `l[0], l[-1] = l[-1], l[0]`
- Ez azt jelenti, hogy megcseréljük a `0`-s és `-1`-es indexű, azaz az első és utolsó elemeket
- a következő lépésben az `1`-es és `-2`-es elemeket cseréljük meg
Ezek magyarázzák is, hogy a `range()`-be miért a hossz felének az egész része került mivel ha:
- a lista páros hosszú, akkor a lista feléig kell haladni, ekkor minden elem meg lesz cserélve
- a lista páratlan hosszú, ekkor a középső elemet nem fogjuk semmivel sem megcserélni, hiszen ezt önmagával megcserélni nem lehet és értelmetlen is.

#### Buborék rendezés Pythonban

A **buborék rendezés** egy egyszerű, de kevésbé hatékony rendezési algoritmus, amely ismételten végigmegy a listán, és ha szükséges, megcseréli a szomszédos elemeket. A legnagyobb elemek fokozatosan "felúsznak" a lista végére, mint a buborékok.

##### Hogyan működik?
1. Az algoritmus **többször végigmegy a listán**.
2. **Összehasonlítja** az egymás melletti elemeket.
3. Ha **rossz sorrendben vannak**, **helyet cserélnek**.
4. A folyamat addig ismétlődik, amíg a lista rendezett nem lesz.

##### Példa kód:

````python
l = [5, 2, 9, 1, 5, 6]

n = len(l)
for i in range(n - 1):  # Ennyi lépés kell a teljes rendezéshez
    for j in range(n - 1 - i):  # Csak a még nem rendezett elemeket vizsgáljuk
        if l[j] > l[j + 1]:  # Ha rossz sorrendben vannak, cseréljük őket
            l[j], l[j + 1] = l[j + 1], l[j]

print(l)  # [1, 2, 5, 5, 6, 9]
````

````python
# De a python maga is tud listákat rendezni
l = [5, 2, 9, 1, 5, 6]
l.sort()
print(l)
````

````python
# Vagy akár fordított sorrendben kiíratni
l = ["Első", "Második", "Harmadik"]
l.reverse()
print(l)
````

#### Listák, mint referenciák
A listáknak, az eddig látott típusokkal ellentétben van egy eltérő működésük.

```python
l1 = [0,1,2]
l2 = l1

l2[0] == 99

```
Vajon mi lesz `l1` 0. indexű elemének értéke? Azt várnánk, hogy ez `0` maradt, hiszen ezt a listát nem módosítottuk, hanem egy másolatot készítettünk róla `l2` változóba, majd ezt módosítottuk. A valódságban viszont `l2` módosításával `l1` is módosult. Ez azért van így, mivel mikor létre hoztuk `l1`-be a listát, valahol a számítógépen létrejött ez a lista, mint adat. A python az `l1` változó értékét összekötötte ezzel az adattal, `l1` változó referrál rá. Amikor azt mondtuk, hogy `l1 = l2`, akkor azt mondtuk, hogy `l2` változó értéke referráljon ugyanerre az adatra. Tehát két változónk van, mindegyiknek egy **referenciája**, de ugyanarra az adatra, tehát mindkét változón keresztül ugyanazt az adatot módosítom.

````python
lista = [1,2,3]
lista_copy = lista

lista_copy[0] = 99
print(lista)
````

Megoldás: **`list.copy()`**

A `copy` használatával egy új adat jött létre az eredeti alapján, `l2` pedig erre az új adatra referrál.

A pythonban általánosan minden ami nem **primitív** típus (`str`, `int`, `float`, `char`, `byte`, `bool`), az olyan módon működik, mint ebben a példában a listák.

#### **Slice**-olás
A slice-olás egy olyan művelet, mellyel listák darabjaiból tudunk új listákat építeni.

A slice-olás másolatot készít, tehát egy slice-olással létrejött listát módosítva az eredeti nem módosul.

Legyegyszerűbb esete a teljes listát vissza adja:

````python
l = [1,2,3]
l_2 = l[:]
print(l_2)

l_2[0] = 99
print(l)
print(l_2)
````

A slice legegyszerűbb felépítése így néz ki:
```python
lista[start:end]
```
szintaktikája hasonlít az indexeléshez, de egy `:` ot is tartalmaz

Az ilyen fromájú slice-olás egy új listát készít `list`-ből, melybe az `start` index és az `end` indexek közti elemeket veszi, belevéve a `start`-adik elemet, de nem az `end`-ediket.
- például a `lista[0:4]` slice-ba belekerül a `0`, `1`, `2` és `3` indexű elemek, de a `4`-es már nem
- Ezt úgy is megfogalmazhatjuk, hogy `start`: az első elem ami még belekerül; `stop`: az első elem ami már nem kerül bele.

Akárcsak az indexelésnél, itt is használhatunk negatív indexeket.

````python
l = [0,1,2,3,4]
print(l[1:3])
````

````python
# amint látjuk, az end már nem kerül bele:
print( l[0:5] )
````

Az első példában láttuk a `[:]` -sliceolást. Ebben az esetben se a `start`, se az `end` nincs megadva.

Akkor, hogy ha azt akarjuk, hogy `start` `0` legyen, vagy `end` `len(l)` legyen, a megfelelő helyre nem kell semmit sem írni. Például:
- `l[0:3]` == `l[:3]`
- `l[3:len(l)]` == `l[3:]`
- `l[0:len(l)]` == `l[:]`

````python
l = list(range(10))
print(l)
````

````python
print(l[0:3])
print(l[:3])
````

````python
print(l[3:len(l)])
print(l[3:])
````

````python
print(l[0:len(l)])
print(l[:])
````

````python
# sliceolásnál is használhatók a negatív indexek
print(l[1:-1])
````

````python
# Az alábbi esetben viszont üres listát kapunk, hisz az utolsó és az első elem között nincs egy elem sem (jobbra haladva)
l[-1:1]
````

#### Slice **step**
A slice-olásnak van egy utolsó, opcionális paramétere, ez pedig a step.

Ez azt határozza meg, hogy mennyivel haladjon `start` és `end` között az a logika amii kigyűjti az elemeket.

A `step`, egy második `:` után kerül leírásra: `l[start:stop:step]`

Tehát az alábbiak ezt jelentik:
- `l[0:len(l):1]` - minden elem egyesével. Ez megegyezik az alábbi slice-al: `l[0:len(l)]`, vagy `l[:]`
- `l[2:8:2]` - minden második elem `2` és `8`-as index közt
- `l[-1:0:-1]` - elemek visszafelé (0. indexű elem nincs benne)
- `l[::-1]` - minden elem visszafele
- `l[2::3]` - minden harmadik elem, a második indextől
- `l[:3:-1]` - minden elem visszafele a 3. indexig (Ebben az esetben a start alapértéke nem 0)

````python
print(l[:])
print(l[::1])
````

````python
print(l[::2]) # minden páros elem
print(l[1::2]) # minden páratlan elem
````

````python
print(l[2:8:2])
````

````python
print(l[-1:0:-1]) # Itt az end 0, azaz a 0-s elemet nem veszi bele
print(l[-1::-1]) # Itt már bele veszi
print(l[::-1]) # és itt is
````

````python
print(l[2::3])
````

````python
print(l[:3:-1])
````

````python
# Vegyük észre, hogy amikor nincs megadva start, ennek értéke 0 lesz, ha a step pozitív és -1, ha a step negatív.
print(l[:5:1])
print(l[0:5:1])

print(l[:5:-1])
print(l[-1:5:-1])
````

````python
# Ugyanígy az end is: 
print(l[5::1])          # Az end nincs megadva, ilyenkor az utolsó elemig akar menni. Mit helyettesít ilyenkor be a stop helyére ha a step pozitív?
print(l[5:len(l):1])    # Mivel 0-tól indexelünk, ezért az utolsó index + 1, ami éppen len(l)-el egyenlő

print(l[5::-1])           # Mit helyettesít be ha a step negatív? Ilyenkor ugy visszafele akar menni az elejéig.
print(l[5:-len(l)-1:-1])  # Mivel a negatív index 1-től kezdődik (ugyanis -0 nincs), ezért a -len(l) a nulladik elemet jelenti, de ha end=0, akkor ezt nem venné bele, ezért ez -len(l)-1 lesz
````

````python
# általánosan
print( l[::1] == l[0:len(l)] )
print( l[::-1] == l[-1:-len(l)-1:-1] )
````

#### List copmrehension
A **`list comprehension`** egy egyszerű szintaktika listák építésére egy ciklus segítségével. Tekintsük az alábbi példát:

````python
# Adjuk meg a számok négyzetét 1-10-ig
szamok = list(range(10))
print(f"számok: {szamok}")

negyzetek = []
for szam in szamok:
    negyzetek.append(szam ** 2)

print(f"négyzetek: {negyzetek}")
````

Ahogy fenti-hez hasonló problémák elég gyakoriak:
- vegyél egy listát
- készíts egy másik listát ennek eleminek valamilyen transzformáltjával
- a transzformáció minden elem esetén ugyanaz

Erre kell legalább három sornyi kód:
- `negyzetek` változó deklarálása
- `for` ciklus ami a listán iterál végig
- `negyzetek`-hez hozzáfűzzük a transzformált értéket

Ez egy kicsit pazarló, ezért van erre egy egyszerűbb szintaktika is, ez pedig a list comprehension:

````python
szamok = list(range(10))
print(szamok)

negyzetek = [szam ** 2 for szam in szamok]
print(negyzetek)
````

Ahogy látni a list comprehension szintaktikájáben lényegében a ciklust tesszük bele a listába úgy, hogy a kifejezés eléggé a beszédre hasonlít:
- vedd a szám négyzetét
- ahol a szám elemek jöjjenek a számok változóból

````python
# Akár így is lehet
print( [szam ** 2 for szam in range(10)] )
````

Ebben a fenti példában a `print` függvénybe egy `*` karaktert tettünk a `sor` kifejezés elé. Ezt az operátort később vesszük, most csak azért van megemlítve, mivel így kevésbé veszi el a figyelmet a hosszas kiíratás, de lényegében ez a karakter az elábbit csinálja:
- `print(*sor)` <=> `print(sor[0], sor[1], sor[2], ... , sor[n])`

Azaz a * operátor szétbontja az n elemű listát n darab paraméterré

Az off topic-ot félre téve a szorzótáblát meg lehet oldani list comprehension-nel is:

````python
szorzotabla = [[a*b for b in range(1,11)] for a in range(1, 11)]

for sor in szorzotabla:
    print(*sor, sep="\t")
````

##### Függvény refaktorálás, szétbontás, függvényből függvény hívás

Dolgozzunk még a korábbi példakódon

A cél itt ugye az volt, hogy kiírjuk egy személy születési évét az életkora alapján. Ragozzuk ezt az évszámot, például:
- 1975-ben
- 1966-ban
- 2010-ben
- 2000-ben
- 1900-ban

````python
# Írassuk ki a "... évben" részt szebben:
#  Ha az utolsó szám 1, 2, 4, 5, 7, 9, akkor mögé tesszük, hogy -ben
#  Ha az utolsó szám 3, 6, 8, akkor mögé tesszük, hogy -ban
#  Ha az utolsó szám nulla, akkor:
#   -ben lesz az utótag 10, 40, 50, 70, 90
#   -ban lesz az utótag 20, 30, 60, 80
#  Ha az utolsó két számjegy nulla:
#   - ben lesz, ha az utolsóü három számjegy nulla - kétezer-ben
#   - egyébként -ban lesz - valahány száz-ban
def print_my_age(name, age):
    born_year = str(datetime.now().year - age) # Kezeljük sztringként
    last_digit = born_year[-1] # Utolsó karakter
    postfix = ""
    if last_digit in "124579":
        postfix = "-ben"
    elif last_digit in "368":
        postfix = "-ban"
    else:
        third_digit = born_year[-2]
        if third_digit in "14579":
            postfix = "-ben"
        elif third_digit != "0":
            postfix = "-ban"
        else:
            second_digit = born_year[-3]
            if second_digit == "0":
                postfix = "-ben"
            else:
                postfix = "-ban"
        
    
    print(f"{name} jelenleg {age} éves, tehát {name} {born_year}{postfix} született.")

name = input("Mi a neved? ")
age = int(input("Hány éves vagy? "))
print_my_age(name, age)
````

Ahogy látjuk a fenti függvény eléggé elbonyolódott a ragozás miatt. Ez a ragozási logika lényegében egy külön működés amit akár több ponton is felhasználhatna a program, és ami önmagában nem kéne, hogy csak az üzenetet kiíró függvény része legyen. Szervezzük ezt ki külön függvénybe

Az új függvényünkben megismerünk egy újabb metódust, ez pedig az `endswith()`. Az `endswith()` metódust sztringeken használhatjuk, megadhatunk neki egy argumentumot, ami egy végződést tartalmaz. A metódus azt mondja meg, hogy valóban ezzel végződik-e a sztringünk. Példa:
- `"almafa".endswith("fa")` -> `True`
- `"almafa".endswith("fa")` -> `True`
- `"almafa".endswith("alma")` -> `False`
- `"almafa".endswith("Fa")` -> `False` mivel számítanak a kis- és nagybetűk
- `"almafa".endswith("zöldalmafa")` -> `False` lesz, de az nem jelent problémát, hogy a `zöldalmafa` hosszabb, mint az `almafa`

````python
def print_number_postfix(number):
    number_as_string = str(number) # mivel ez a függvény int-ként kapja meg a számot
    postfix = ""

    last_digit = number_as_string[-1]
    if last_digit in "124579":
        postfix = "-ben"
    elif last_digit in "368":
        postfix = "-ban"
    # Ez a függvény általánosan működik egy számra, így akár kaphatna 1 jegyű számot is. Ekkor nincs utolsó előtti számjegy, így a raggal a szám "0-ban" lesz.
    elif len(number_as_string) == 1:
        postfix = "-ban"
    else:
        penultimate_digit = number_as_string[-2]
        if penultimate_digit in "14579": # 10-ben, 40-ben, 50-ben, 70-ben, 90-ben
            postfix = "-ben"
        elif penultimate_digit != "0": # 20-ban, 30-ban, 60-ban, 80-ban
            postfix = "-ban"
        else:  # Azt tudjuk, hogy ha az utolsó két számjegy 0, akkor van utolsó előtti-előtti számjegy is, hiszen az int(00)-ból 0 lesz, tehát önmagában két nulla nem lehetséges
            penultimate_before_last_digit = number_as_string[-3]
            if penultimate_before_last_digit != "0": # valahán-száz-ban
                postfix = "-ban"
            elif number_as_string.endswith("000000"): # Ha legalább 6 nullával végződik a szám, akkor az valahány millió, milliárd, billió... Ezek posztfixe mindig -ban
                postfix = "-ban"
            else: # Más esetben legalább három és legfeljebb öt nullára végződik a szám, ami kimondva valahány ezer -> a postfix: -ben lesz
                postfix = "-ben"
    print(postfix, end="")

def print_a_number_with_postfix(number):
    print(number, end="")
    print_number_postfix(number)
    print() # A végén egy üres sort printelünk
                

print_a_number_with_postfix(0)
print_a_number_with_postfix(11)
print_a_number_with_postfix(13)
print_a_number_with_postfix(110)
print_a_number_with_postfix(120)
print_a_number_with_postfix(500)
print_a_number_with_postfix(1000)
print_a_number_with_postfix(123000)
print_a_number_with_postfix(4000000)
````

Ahogy láttuk, egy függvényből meghívhatunk egy másikat is. Bár azt már tanultuk, hogy ha a programkódban meghívunk egy függvényt, akkor azt előbb definiálni kell. Ez a függvények esetén ennél rugalmasabb. A példában van két függvényünk, `func_a` és `func_b`. `func_a` meghívja `func_b`-t. Ebben az esetben nem probléma, ha előbb azt definiáljuk, amely a másikat meghívja, ha később valahol definiáljuk a másikat:

````python
def func_a():
    print("meghívódott func_a")
    func_b()
    print("lefutott func_a")

def func_b():
    print("meghívódott func_b")
    print("lefutott func_b")

func_a()
````

A fenti eset azért működik, mivel mikor meghívjuk `func_a` függvényt az utolsó sorban, addigra már definiálva van `func_b` is. Tehát mire a program futása eléri azt a pillanatot, mikor `func_b`-t meg kell hívni, addigra már lefutott `func_b` definiálása.

Tekintsük az alábbi ellenpéldát:
- ebben `func_c` meghívja `func_d`-t. Viszont `func_c`-t még azelőtt hívjuk meg, hogy `func_d` definiálása lefutna. Nem meglepő, hogy ez hibát eredményez:

#### Függvények scope-ja
Ahogy az korábban már említve lett, egy függvény belső változóit kívülről nem látjuk. Fordított esetben viszont működhet aa dolog:

````python
def my_function():
    print(variable)

variable = "alma"
my_function()
````

Ez azért működik, mivel a függvény rálát azokra a változókra, amelyek a hívás előtt jöttek létre. Szakszóval kifejezve, a `variable`-t tartalmazza a függvény scope-ja. 

A függvény scope-ja rejtett a függvényen kívül kód előtt, azaz a függvényen kívüli kódokra a függvény közvetlenül nincs ráhatással. 

Ha rálát a függvény a `variable` változóra tudja is azt módosítani?

````python
def my_function():
    # print(variable)
    variable = "körte"

variable = "alma"
my_function()

print(variable)
````

Úgy látszik, hogy nem. De akkor hogyhogy ki tudja írni?

A választ megkaphatjuk, ha egyszerre akarjuk kiírni és módosítani is:

````python
def my_function():
    print(variable)
    variable = "körte"

variable = "alma"
my_function()

print(variable)
````

Itt hibát kaptunk. Miért történt ez?

A függvények scope-jának tekintetében kétféle változót különböztetünk meg:
- **local**: a függvényen belül létezik csak
- **global**: a függvényen kívül létezik

Amikor egy függvényben hivakozunk egy változóra (például `print(variable)`), akkor a python megnézi, hogy ez szerepel-e a lokálisan definiált változók közt. Ha nem szerpel, akkor a globális változók közt próbálja megkeresni. 

A python mikor egy függvényt futtat, többször is végighalad a kódon és értelmezi azt. Például kigyűjti, hogy milyen lokális változók lesznek definiálva a kódban. Amikor lefuttatja a kódot és történik egy hivatkozás egy változóra, amelynek nevéről megállapította, hogy az egy lokális változó lesz a kódban, akkor először is azt vizsgálja meg, hogy létezik-e már a változó, definiáltuk-e. Amennyiben még nem, nem fogja megnézni, hogy a globális változók közt szerepel-e, hiszen egy változó csak egy scope-ból jöhet, így hibát kapunk.

````python
def my_function():
    print(variable)
    if False:
        variable = "körte"

variable = "alma"
my_function()
````

Ezt bizonyítja a fenti kód is. Az `if False:` ág nem képes lefutni sohasem, viszont a benne található változó deklaráció már beszorította a `variable` változót a függvény lokális scope-jába, így a `print` függvény ezt nem fogja megtalálni, mivel a függvényben nincs deklarálva a változó sehol sem.

De akkor mégis hogyan módosítunk külső változót? 

A pythonban a `global` kulcsszó segítségével mondhatjuk meg egy függvényben használt változóról, hogy annak a scope-ja globális legyen:

````python
def my_function():
    global variable
    print(variable)
    variable = "körte"

variable = "alma"
my_function()

print(variable)
````

Mi történt?
- létrehoztunk egy globális változót `variable` néven, "alma" értékkel
- mehívtuk a `my_function()` függvényt
- a függvény azt mondta, hogy ő használni fogja a globális `variable` változót
- kiíratta ennek az értékét
- felül írta az értékét a "körte" sztringgel
- a függvény futása véget ért
- kiíratva a globális `variable` változót, azt tapasztaltuk, hogy ennek értéke valóban megváltozott "körte"-re

A fentiekhez még hozzá tartozik, hogy bár a `global` kulcsszó nélkül egy globális változóhoz nem lehet új értéket rendelni, de ennek háttéradatát tudjuk módosítani, mondjuk egy lista esetén.

````python
def my_function():
    print(variable)
    variable.append(5)

variable = [1, 2, 3, 4]
my_function()

print(variable)
````

Ez azért lehetséges, hiszen nem a változóhoz akarunk új értéket rendelni, hanem az értéken végzünk módosítást

````python
# több globális változó esetén:
def my_function():
    global a, b, c
    a, b, c = c, b, a
a = 1
b = 2
c = 3

print(a, b, c)
my_function()
print(a, b, c)
````

##### Pitagorasz háromszög:

A háromszög derékszögű, ha:

 $a^2+b^2=c^2$

 , ahol $a$ és $b$ a befogók, $c$ pedig az átfogó, azaz a leghosszabb oldal.

 Írjuk meg úgy a függvényt, hogy nem feltétlen lesz c a leghosszabb oldal.

````python
def is_a_right_triangle(a, b, c):
    if a ** 2 + b ** 2 == c ** 2:
        return True
    if a ** 2 + c ** 2 == b ** 2:
        return True
    if b ** 2 + c ** 2 == a ** 2:
        return True
    return False

print(is_a_right_triangle(3,4,5))
print(is_a_right_triangle(5,4,3))
print(is_a_right_triangle(3,5,4))
print(is_a_right_triangle(1,1,2))
````

##### Fibonacci számok

az algoritmus egyszerű:
- az első szám 0
- a második szám 1
- a harmadik szám az első két szám összege = 1
- a negyedik szám az előző két szám összege = 2
- 3
- 4

Általánosan:

$F_0 = 0, \quad F_1 = 1$

$F_n = F_{n-1} + F_{n-2}$

````python
def fibonacci(n):
    fibonacci_numbers = [0, 1]

    while len(fibonacci_numbers) < n:
        fibonacci_numbers.append(fibonacci_numbers[-1] + fibonacci_numbers[-2])
    
    return fibonacci_numbers[n-1]

for i in range(10):
    print(fibonacci(i))
````

- Ez egyfajta ciklikus működést eredményez -> Mindenképp kell valamiféle feltétel ami megszakítja a ciklikusságot. A fenti példában az szekítja meg, ha egy listában nincsenek további listák
- Másfelől egyfajta többszörös elágazódás is rejlik benne, egyféle fa struktúra. Jellemzően a fa-struktúrák, változó mélységű hierarchiák feldolgozása, bejárása és építése valamilyen rekurziót tartalmaz.

Könnyen megoldhatjuk a korábbi fibonacci számos megoldást rekurzívan is:

````python
def fibonacci(n):
    if n < 1:           # A fibonacci sorozatnak (mint bármely más sorozatnak) nincs 0., vagy -1. eleme
        return None
    if n <= 2:          # A fibonacci sorozat első két eleme 1
        return 1
    else:
        return fibonacci(n -1) + fibonacci(n -2)  # Minden más esetben a sorozat képletéből indulunk ki F(n) = F(n-1) + F(n-2)
    
print(fibonacci(5))
````

A fenti megoldás nem túl hatékony. Vegyük példaként az ötödik fibonacci szám kiszámítását:
- $fibonacci(5) = fibonacci(4) + fibonacci(3)$
    - $fibonacci(4) = fibonacci(3) + fibonacci(2)$
        - $fibonacci(3) = fibonacci(2) + fibonacci(1)$
            - $fibonacci(2) = 1$
            - $fibonacci(1) = 1$
        - $fibonacci(2) = 1$
    - $fibonacci(3) = fibonacci(2) + fibonacci(1)$
        - $fibonacci(2) = 1$
        - $fibonacci(1) = 1$

-> $fibonacci(5) = ((1 + 1) + 1) + (1 + 1)$

Ahogy láttuk az adott értékeket ennyiszer számoltuk ki:
- fibonacci(5) értékét egyszer
- fibonacci(4) értékét egyszer
- fibonacci(3) értékét kétszer
- fibonacci(2) értékét háromszor
- fibonacci(1) értékét kétszer 

Ebben az esetben 4 felesleges lépést tettünk. Nem minden esetben hasznos a rekurzió. Ha egy függvény egyetlen futása során a törzsében kétszer is meghívja önmagát, akkor lehetséges, hogy felesleges lépéseket teszünk

````python
import time
def fibonacci_with_precalculation(n):
    fibonacci_numbers = [1, 1]

    while len(fibonacci_numbers) < n:
        fibonacci_numbers.append(fibonacci_numbers[-1] + fibonacci_numbers[-2])
    
    return fibonacci_numbers[n-1]

def fibonacci_with_recursion(n):
    if n < 1:
        return None
    if n <= 2:
        return 1
    else:
        return fibonacci(n -1) + fibonacci(n -2)

n = 38

start = time.time() # aktuális idő lekérés
print(fibonacci_with_precalculation(n))
stop = time.time() # aktuális idő futás után
dt_using_precalculation = stop - start # ez egy float érték lesz
# Az f-stringben a behelyettesített érték mögött álló ":.2f" azt jelenti, hogy a float 2-tizedes jegyre legyen kerekítve
print(f"Előszámolós megoldás sikeresen lefutott, n = {n} esetén {dt_using_precalculation:.2f}s alatt ")

start = time.time() # aktuális idő lekérés
print(fibonacci_with_recursion(n))
stop = time.time() # aktuális idő futás után
dt_using_recursion = stop - start  # Futás során eltelt időt kiszámolhatjuk a két időpillanat különbségével
print(f"Rekurzív megoldás sikeresen lefutott, n = {n} esetén {dt_using_recursion:.2f}s alatt ")
````

A `*`-os változónak nem kell a végén szerepelnie. Amennyiben nem a végén szerepel:

    - `d` értéke `my_tuple[-3]` lesz
    - Ahogy látjuk, ami maradt az a `my_tuple[2:-3]`, ezek az értékek lesznek `c` változóhoz rendelve.

A fentiek miatt láthatjuk, hogy egy ilyen unpacking-be csak egy `*`-os kifejezés kerülhet, máskülönben nem egyértelmű, hogy hogyan osszuk szét a változókat.

A függvények csak egy ilyen több értékes paramétert kaphatnak, melynek értéke egy tuple lesz, amibe több érték kerülhet, de akár üresen is maradhat. Készítsünk egy függvény(eke)t, kártyás vásárlások szimulálására.
- a függvény vár egy egyenleget, valamennyi megvárárolandó termék árát, és egy devizanemet, amely opcionális
- Ha a vásárlónak van fedezet a kártyáján, írjuk ki, hogy a várárlás sikeres volt, majd írjuk ki az új egyenleget
- Amennyiben a vásárlás sikertelen, írjuk ezt ki, illetve, hogy mennyi pénz hiányzik még.
- Ha viszont nincs megvásárolt termék, a tranzakció érvénytelen

````python
def print_transaction_result(balance, *prices, currency="HUF"):
    sum_expenses = sum(prices)                                      # a sum() függvény összeadja egy szekvencia elemeit, amennyiben azok számok (egyébként hibát dob)
    remaining_balance = balance - sum_expenses                      # mennyi pénz marad a kártyán ha kifizetné a termékeket
    if not len(prices):                                             # Az "if <szekvencia> kifejezésnél a szekvencia bool-lá konvertálódik. Ez akkor lesz False, ha üres, ha van benne elem True
        print(f"Érvénytelen tranzakció: Nincs összeg!")
    if remaining_balance < 0:
        print(f"Sikertelen vásárlás: Nincs elég fedezet a kártyán. Hiányzó összeg: {-remaining_balance} {currency}")
    else:
        print(f"Sikeres vásárlás! Fennmaradó egyenleg: {remaining_balance} {currency}")

print_transaction_result(20000, 200, 454, 50, 1299)
print_transaction_result(1291.23, 0.99, 31, 7.83, currency="EUR")
print_transaction_result(1500)
print_transaction_result(12000, 1500, 6900, 3355, 789)
````

#### Dictionary-k
A dictionary egy python adatstruktúra. Bár nem szekvencia, de sok, a szekvenciákra jellemző műveletet támogat.

A dictionary-k nevükhöz hasonlóan valóban szótár-szerűek. Ha például magyar "tyúk" szó angol megfelelőjét akarjuk megtalálni, egy szóárban könnyen megkereshetjük ezt, mivel a szótárban a magyar-angol szavak össze vannak párosítva.

Az analógia másik oldalát tekintve a szó amit keresünk, az a kulcs avagy `key`, a fordítás pedig az érték, azaz `value`. Emaitt a dictionary-ket **kulcs-érték**, vagy **key-value** tárolóknak is hívjuk.

A dictionary tulajdonságai:
- A kulcs egyedi - akárcsak egy szótárban. Az "ősz" kulcs egyszer szerepel, ha ennek több jelentése van, mint "fall", vagy "gray", akkor azok az "ősz" kulcs alatt vannak listázva
- Az érték nem kell, hogy egyedi legyen. Mint a szótárban is, a "Fall" szó tartozhat az "ősz" és a "zuhanás" kulcsokhoz is.
- A kulcs bármilyen primitív típus lehet. Ilyenek a sztring, az int, a float, vagy bool is, de összetett típusú, mint mondjuk lista nem lehet.
- a `len()` függvény a dictionary-k esetén a kulcs-érték párok számát adja vissza
- A dictionary-kből kulcs alapján lekérdezhető az érték (mivel a kulcs egyedi), fordítva viszont nem működik (mivel a value nem feltétlen egyedi)

##### Dictionary-k létrehozása:
```py
    my_dictionary = {"alma": "apple", "ablak": "window", "ősz": ["fall", "grey"], "zuhanás": "fall"}
```
Ahogy azt látni:
- a dictionary-ket `{}` kunkori-zárójelpár között tudjuk definiálni
- egy kulcsot és annak értékét `:` karakterrel választjuk el egymástól
- a kulcs-érték párokat, azaz az elemeket pedig szokász szerint `,` karakterrel

````python
dictionary = {"alma": "apple", "ablak": "window", "ősz": ["fall", "grey"], "zuhanás": "fall"}
account_balance = {"Varmanek Borbála": 2000000, "Pintér Miklós": 380000000, "John Lebovski": 7600000}
empty_dict = {}

print(dictionary)
print(account_balance)
print(empty_dict)
````

##### Elemek lekérése
A dictionary-k elemeit, egy indexeléshez hasonló szintaxissal kaphatjuk meg:

````python
dictionary = {"alma": "apple", "ablak": "window", "ősz": ["fall", "grey"], "zuhanás": "fall"}

print(dictionary["alma"])
print(dictionary["ősz"])
````

Ha nem létező kulcsra hivatkozunk, error-t kapunk:

````python
print(dictionary["nem létező kulcs"])
````

Az `in` és `not in` kulcsszavakkal azt tudjuk ellenőrizni, hogy egy érték szerepel-e a kulcsok között. Azt viszont nem, hogy szerepel-e a value-k közt:

````python
dictionary = {"alma": "apple", "ablak": "window", "ősz": ["fall", "grey"], "zuhanás": "fall"}

print("alma" in dictionary)               # Igen
print("nem létező kulcs" in dictionary)   # Nem, hiszen nincs ilyen kulcs
print("apple" in dictionary)              # Nem, hiszen ez nem egy kulcs
````

##### Értékadás, felülírás
Érték felülírása a listákéhoz hasonló, de ugyanígy adhatunk hozzá értékeket:

````python
# Hozzáadjuk a "körte" kulcsot "bulb" értékkel
dictionary["körte"] = "bulb"
print(dictionary["körte"])

# Felülírjuk a "körte" kulcsot "pear" értékkel
dictionary["körte"] = "pear"
print(dictionary["körte"])
````

A **`setdefault()`** metódussal is hozzá adhatunk értéket a dict-ünkhöz, viszont ha a kulcs már nem szerepel a dict-ben, akkor ez nem írja felül

````python
# A dict-ben van egy "a" kulcs, ennek értéke 999
d = {"a": 999}
# Ha még "a" kulcs nem szerepel a dict-ben, kerüljön bele 0-s értékkel
d.setdefault("a", 0)
# Ha még "b" kulcs nem szerepel a dict-ben, kerüljön bele 0-s értékkel
d.setdefault("b", 0)

# Ahogy látni fogjuk, "a" értéke meg maradt 999, míg "b" belekerült a dict-be 0-s értékkel
print(d)
````

##### Iterálás
Ha iterálunk, az a kulcsokon fog iterálni:

````python
# Így is formázhatjuk, hogy átláthatóbb legyen:
dictionary = {
              "alma": "apple",
              "ablak": "window",
              "ősz": ["fall", "grey"],
              "zuhanás": "fall"
              }

for key in dictionary:
    print(key)
````

````python
# De persze le tudjuk kérni a kulcs alapján az értéket
for key in dictionary:
    print(dictionary[key])
````

##### dict.items()
Ha a kulcsot és az értéket is szeretnénk használni, akkor iteráljuk a `dictionary.items()`-en végig

````python
for key, value in dictionary.items():
    print(f"kulcs: {key};\t\térték: {value}")
````

Ez hogyan is működött? Nézzük meg, mi az a `dict.items()`:

````python
print(dictionary.items())
````

Habár a "dict_items" felirat összezavaró lehet, de azt látjuk, hogy a belső struktúra tuple-ökben tárolja a kulcs-érték párokat.

A `dict.items()` egy olyan metódus, mely a dict-ből egy generátort állít elő. A generátor objektumok (ilyen a `range` függvény visszatérési értéke) iterálhatóak, de nem indexelhetőek. Ez azért van, mert ahogy a neve is sugallja, a generátor nem tartalmazza az elemeket, hanem kigenerálja, ráadásul ezt csak sorrendben képes megtenni. Ezt valahogy úgy kell elképzelni, hogy egy generátor-nak mondhatom azt, hogy "add ide a következő elemet" és ekkor ezt legenerálja nekem, de innentől azt már nem tudja vissza adni. Kicsit olyan ez, mint egy sor a kasszánál. A kasszás csak a sorban következő embert tudja oda hívni, elintézik amit el akarnak, majd a vásárló elmegy. A kasszás nem tudja a sorban harmadik embert szólítani, és a kettővel ezelőtt lévő vásárlót sem híhatja vissza.

Ez kicsit összezavaró, hiszen mikor kiírattuk a generátor értékét, láttuk benne az elemeket. Ez azért van, mert maga a kiíratás generáltatja ki az összes elemet.

A generátorokat, viszont listává konvertálhatjuk, hogy láthassuk mik is vannak benne:

````python
print(list(dictionary.items()))
print(list(dictionary.items())[0])
````

Szóval az `dictionary.items()`-en való iterálás elemei a `key` és `value`-ból álló párok két elemű tuple-jei lesznek. 

Ebből jól látszik, hogy mikor azt mondjuk, hogy `for key, value in dictionry.items()`, ott a `key` és `value` ciklus változók a tuple két eleme lesznek.

````python
tuple_dict = {"a": (1,2), "b": (3,4), "c": (5,6)}

for key, val_1, val_2 in tuple_dict.items():
    print(key, val_1, val_2)
````

Miért történt hiba? Írassuk ki a `tuple_dict.items()` egy elemét:

````python
print(list(tuple_dict.items())[0])
````

Egyértelmű:
- Az items első elem a kulcs
- második eleme a value
- a value pedig maga is egy tuple
- tehát ez egy tuple a tuple-ben

De hogyan kellene ezt kezelni? Azt tudjuk már, hogy nem így:

````python
my_tuple = (1, 2)
my_outer_tuple = ("a", my_tuple)
print(my_outer_tuple)

a, b, c = my_outer_tuple
print(a)
print(b)
print(c)
````

Hanem így:

````python
my_tuple = (1, 2)
my_outer_tuple = ("a", my_tuple)
print(my_outer_tuple)

a, (b, c) = my_outer_tuple
print(a)
print(b)
print(c)
````

Azzal, hogy zárójelbe tettük `b` és `c` változót, a pythonnak azt mondtuk, hogy:
- `a` értéke legyen az `outer_tuple[0]` elem
- `(b, c)` értéke legyen az `outer_tuple[1]` elem (jelentsen ez bármit is). Ez annyit tesz, hogy
    - `b` értéke legyen az `outer_tuple[1][0]` elem
    - `c` értéke pedig az `outer_tuple[1][1]` elem

Mi a probléma akkor a korábbi megoldással?
- `a, b, c = outer_tuple` a pythonnak annyit tesz, hogy:
    - `a` értéke legyen az `outer_tuple[0]` elem
    - `b` értéke legyen az `outer_tuple[1]` elem
    - `c` értéke legyen az `outer_tuple[2]` elem -> Ez nem létezik, error-t fog dobni.

Tehát a ciklus helyesen:    

````python
tuple_dict = {"a": (1,2), "b": (3,4), "c": (5,6)}

for key, (val_1, val_2) in tuple_dict.items():
    print(key, val_1, val_2)
````

##### A dictionary metódusai:
- `clear`: töröl minden elemet
- `keys` : vissza adja a kulcsokat
- `values`: csak az értékeket adja vissza
- `get`: alapvetően úgy működik, mint a `dict["kulcs"]` érték lekérés, de van némi csavar (kifejtve később)

##### A `get` használata kétféle lehet:
- `dict.get("some_key")`: visszaadja a "some_key" kulcshoz tartozó értéket. Ha ez nincs benne, akkor `None`-al tér vissza. (Ez eltér a `dict["some_key"])` művelettől, ami itt error-t dobna)
- `dict.get("some_key", "some_default_value")`: ugyanúgy működik, mint a felső, csak hiányzó kulcs esetén a "some_default_value" értékkel tér vissza

````python
print(dictionary.get("alma"))
print(dictionary.get("alma", "HIÁNYZÓ"))
print(dictionary.get("hiáynzó kulcs"))
print(dictionary.get("hiáynzó kulcs", "HIÁNYZÓ"))
````

````python
for value in dictionary.values():
    print(value)
````

##### Elemek törlése

4 féle módon történhet:
- `dict.pop("kulcs")` - törli az adott kulcsú elemet
- `dict.popitem()` - törli az utolsó elemet
- `del dict["kulcs"]` - törli az adott kulcsú elemet (nem ez a bevett módja, a pop() az elterjedtebb )
- `dict.clear()` - törli az összes elemet

Ezek közül jellemzően a `.pop()`-ot használjuk

````python
d = {"a": 2}
print("a" in d)
popped = d.pop("a")
print("a" in d)

print(popped)
````

##### Referenciák

Akár a listáknál, úgy a dict-eknél is probléma lehet referencia kérdés. A megoldás itt is a `.copy()` metódus:

````python
dict_1 = {"a": 1, "b": 2}
dict_2 = dict_1

print(dict_1)
dict_2["c"] = 3
print(dict_1)
print(dict_2)
````

````python
dict_1 = {"a": 1, "b": 2}
dict_2 = dict_1.copy()

print(dict_1)
dict_2["c"] = 3
print(dict_1)
print(dict_2)
````

##### Dict comprehension
Dicteknél is lehetőségünk van egysoros dict építésre for ciklus alapján, a szintaktika is nagyon hasonló annyi különbséggel, hogy:
- természetesen az egész kifejezés `{}` kunkori-zárójelpár közé kerül
- és minden elem esetén a kulcsot és az értéket is meg kell adni

Példa kódok:

````python
items = [("a", 1), ("b", 2), ("c", 3), ("d", 4)]

my_dict = {}
for key, value in items:
    my_dict[key] = value

print(my_dict)
````

````python
items = [("a", 1), ("b", 2), ("c", 3), ("d", 4)]
my_dict = {item[0]: item[1] for item in items}
print(my_dict)
````

````python
items = [("a", 1), ("b", 2), ("c", 3), ("d", 4)]
my_dict = {key: value for key, value in items}
print(my_dict)
````

Itt egy életszagúbb példa

Megjegyzés: a sztringek ".split()" metódusa egy karakter mentén darabolja a sztringet egy listányi sztringgé:

````python
items = "a:alma, b:körte, c:barack, d:cékla"
my_dict = {item.split(":")[0]: item.split(":")[1] for item in items.split(", ")}
print(my_dict)
````

#### zip

````python
keys = ["a", "b", "c", "d"]
values = [1, 2, 3, 4]

items = list(zip(keys, values))
print(items)
print(items[0])
````

````python
my_dict = {key: value for key, value in zip(keys, values)}
print(my_dict)
````
